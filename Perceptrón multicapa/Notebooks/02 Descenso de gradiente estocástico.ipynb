##Dataset

```python
import numpy as np
from random import random
import matplotlib.pyplot as plt
from IPython.display import clear_output
```

```python
NUM_PUNTOS = 100

x = list()
y = list()

for _ in range(NUM_PUNTOS):
  coordenada_x = random()
  coordenada_y = np.sin(2 * np.pi * coordenada_x)
  ruido = random() - 0.5 # (-0.5,0.5)
  coordenada_y += ruido
  x.append( coordenada_x )
  y.append( coordenada_y )
# otra opción con numpy: coordenada_x = np.random.rand(NUM_PUNTOS)

plt.plot(x, y, 'ro')
plt.show()

print("\nAlgunos puntos:\n")
for i in range(5):
  print(f'({x[i]}, {y[i]})')
```

```python
x2 = np.arange(0, 1, 0.01)
y2 = np.sin(2 * np.pi * x2)

plt.plot(x, y, 'ro', x2, y2, 'b-')
plt.show()
```

##Polynomial Regression

```python
def eval_poly(w, x):
  h = 0
  for j in range(len(w)):
    h = h + w[j] * (x**j)
  return h
```

```python
D = 3
alpha = 0.1
epochs = 2000

w = list()
for _ in range(D+1):
  w.append(random() - 0.5)
# otra opción con numpy: w = np.random.rand(D) - 0.5

modelo_x = np.linspace(0, 1, NUM_PUNTOS)
H = list()

for i in range(NUM_PUNTOS):
  H.append( eval_poly(w, modelo_x[i]) )
plt.plot(modelo_x, H, 'g-', x2, y2, 'b-')
plt.show()
```

```python
for e in range(epochs):

  for i in range(NUM_PUNTOS):
    for j in range(D+1):
      w[j] = w[j] + alpha * (y[i] - eval_poly(w, x[i])) * (x[i]**j)

  for i in range(NUM_PUNTOS):
    H[i] = eval_poly(w, modelo_x[i])

  if e%50 == 0:
    clear_output(wait=True)
    plt.plot(modelo_x, H, 'g-', x2, y2, 'b-')
    plt.show()
```
