##Funciones

```python
def cuadrado(x):
  c = x*x
  return c

print(cuadrado(5))

def cuadrado_cubo(x):
  c = x*x
  u = x*x*x
  return c, u

cuad, cubo = cuadrado_cubo(3)
print(cuad, cubo)
```

##Listas

```python
semana = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']
print(semana)

fin_de_semana = list()
print(fin_de_semana)

fin_de_semana.append('Sábado')
print(fin_de_semana)

fin_de_semana.append('Domingo')
print(fin_de_semana)

entre_semana = list()
for dia in semana:
  if dia not in fin_de_semana:
    entre_semana.append(dia)
print(entre_semana)

numeros = [2,3,6,4,8,5,7,3,2]
print(numeros)

numeros.remove(8)
print(numeros)

menor_mayor = sorted(numeros)
mayor_menor = sorted(numeros, reverse=True)
print(menor_mayor)
print(mayor_menor)
```

##Números aleatorios

```python
from random import random

for _ in range(10):
	val = random()
	print(val)

from random import randint

for _ in range(10):
	val = randint(0, 10)
	print(val)

from numpy.random import rand

val = rand(10)
print(val)

from random import uniform

for _ in range(10):
 val = uniform(-10, 10)
 print(val)

import numpy as np

val = np.random.uniform(-1,0,10)
print(val)
```

##Matplotlib


```python
import matplotlib.pyplot as plt

x = np.linspace(0,1,100)
print(x)
y = np.sin(2*np.pi*x)

x2 = np.arange(0, 1, 0.02)
print(x2)
y2 = np.cos(3*np.pi*x2)

plt.figure(figsize=(7, 5), constrained_layout=True)
plt.plot(x, y, color='red', label='Seno(2 Pi x)')
plt.plot(x2, y2, color='blue', label='Coseno(3 Pi x)')
plt.scatter(x, y)
plt.scatter(x2, y2)
plt.xlabel("x")
plt.ylabel("Función trigonométrica")
plt.title("Una gráfica básica")
plt.legend();
```
